```python
import zipfile  # import zipfile

# Open the ZIP and read the CSV directly without extracting files.
zip_path = r"C:\Users\HP\Downloads\NGA_2025_HFCP_v01_M_CSV.zip"
with zipfile.ZipFile(zip_path) as z:
    print(z.namelist())
    with z.open("level2_price_garri.csv") as f:
        df = pd.read_csv(f)
```
